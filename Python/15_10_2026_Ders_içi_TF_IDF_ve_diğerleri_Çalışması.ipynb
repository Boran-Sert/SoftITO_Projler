```python
import math
import sys
import warnings
warnings.filterwarnings('ignore')
```

```python
cumleler = [
"İstanbul'da sabahları trafik çok yoğun oluyor.",
"Yoğun trafik yüzünden birçok kişi işe metroyla gidiyor.",
"Metro, İstanbul'da en hızlı ulaşım seçeneklerinden biri.",
"Vapur ise boğazı geçmek için keyifli bir ulaşım yolu.",
"Sabahları vapurda çay içmek İstanbullular için bir alışkanlık.",
"Hafta sonları trafik azalıyor ve şehir daha sakin oluyor.",
"Sakin bir hafta sonunda boğaz kıyısında yürüyüş yapmak çok güzel.",
]
```

```python
def normalize(text):
    translation_table = str.maketrans("çğıöşüÇĞİÖŞÜ", "cgiosucgiosu")
    return text.translate(translation_table)
```

```python
normalize_cumleler = []
for cumle in cumleler:
    normalize_cumleler.append(normalize(cumle))
```

```python
def tokenize(text):
    return text.lower().split()
```

```python
tokenize_kucuk = [tokenize(d) for d in normalize_cumleler]
print("Tokenize edilmiş dokümanlar:")
for i, d in enumerate(tokenize_kucuk):
    print(f"  Doküman {i+1}: {d}")
```

```python
from sklearn.feature_extraction.text import TfidfVectorizer
import pandas as pd

vectorizer = TfidfVectorizer()
tfidf_matrisi = vectorizer.fit_transform(normalize_cumleler)

df = pd.DataFrame(
    tfidf_matrisi.toarray(),
    columns=vectorizer.get_feature_names_out(),
    index=[f"Doküman {i+1}" for i in range(len(normalize_cumleler))]
)
df
```

```python
print("Her Dokümanda En Yüksek TF-IDF Skoruna Sahip Kelime:\n")
for i, doc in enumerate(normalize_cumleler):
    satir = df.iloc[i]
    en_iyi = satir.idxmax()
    skor = satir.max()
    print(f"Doküman {i+1}: '{doc}'")
    print(f"  -> En önemli kelime: '{en_iyi}' (skor: {skor:.4f})\n")
```

```python
# ÖRNEK: Anlamsal ilişkinin TF-IDF ile yakalanamaması
print("=" * 70)
print("PROBLEM 1: Anlamsal İlişki (Semantics)")
print("=" * 70)

def tf(term, tokenize_kucuk):
    count = tokenize_kucuk.count(term)
    return count / len(tokenize_kucuk) if len(tokenize_kucuk) > 0 else 0

def idf(term, tokenize_kucuk):
    doc_count = sum(1 for doc in tokenize_kucuk if term in doc)

    return math.log(len(tokenize_kucuk) /  doc_count)

print("TF-IDF Değerleri:\n")
for i, doc in enumerate(tokenize_kucuk):
    print(f"--- Doküman {i+1}: '{cumleler[i]}' ---")
    for term in sorted(set(doc)):
        tf_val = tf(term, doc)
        idf_val = idf(term, tokenize_kucuk)
        tfidf_val = tf_val * idf_val
        print(f"  '{term}' -> TF: {tf_val:.4f}, IDF: {idf_val:.4f}, TF-IDF: {tfidf_val:.4f}")
    print()
```
